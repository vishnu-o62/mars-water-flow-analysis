```
import rasterio
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# Load DEM
with rasterio.open("../data/raw/hirise/ESP_031871_1985.jpg") as src:
    dem = src.read(1)

# Create figure
plt.figure(figsize=(12, 8))

# Plot DEM
img = plt.imshow(dem, cmap='terrain')

# Colorbar
cbar = plt.colorbar(img)
cbar.set_label("Elevation (meters)")

# Custom labels for elevation ranges
cbar.set_ticks([25, 75, 125, 175, 225])
cbar.set_ticklabels([
    "Very Low",
    "Low",
    "Moderate",
    "High",
    "Very High"
])

# Title and axes labels
plt.title("MOLA DEM with Terrain Classification")
plt.xlabel("X Coordinate (pixels)")
plt.ylabel("Y Coordinate (pixels)")

# Create legend
legend_elements = [
    Patch(facecolor='blue', label='Very Low Elevation (Potential Channels)'),
    Patch(facecolor='cyan', label='Low Elevation (Valleys)'),
    Patch(facecolor='limegreen', label='Moderate Elevation (Slopes)'),
    Patch(facecolor='gold', label='High Elevation (Plateaus)'),
    Patch(facecolor='white', edgecolor='black',
          label='Very High Elevation (Ridges)')
]

plt.legend(
    handles=legend_elements,
    loc='lower center',
    bbox_to_anchor=(0.5, -0.3),
    ncol=1,
    fontsize=10
)

# Additional explanation
plt.figtext(
    0.15,
    -0.15,
    "Blue : Very Low Elevation (Potential Channels)\n"
    "Cyan : Low Elevation (Valleys)\n"
    "Green : Moderate Elevation (Slopes)\n"
    "Yellow : High Elevation (Plateaus)\n"
    "White : Very High Elevation (Ridges)",
    fontsize=10
)
plt.tight_layout()
plt.show()
```
